```python
#Histogram
print("Mithunjai.E\t\tDSV - 29_01_26\t\t24BAD071")
import pandas as pd
import matplotlib.pyplot as plt

data = "/content/29-01-2026.csv";

df = pd.read_csv("/content/29-01-2026.csv")

happiness_score = df['HappinessScore'].head(50)

plt.hist(happiness_score, bins = 4)
plt.title('Histogram Happiness Scores')
plt.show()
```

```python
#box plot
print("Mithunjai.E\t\tDSV - 29_01_26\t\t24BAD071")
import pandas as pd
import seaborn as sns

data = "/content/29-01-2026.csv";

df = pd.read_csv(data)

sns.boxplot(y='Country',x='HappinessScore',data=df.head(20))
plt.title('BOX plot')
plt.show()
```

```python
#Scatter plot
print("Mithunjai.E\t\tDSV - 29_01_26\t\t24BAD071")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/29-01-2026.csv").head(25)

plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Log GDP per capita',
    y='HappinessScore',
    data=df
)

plt.title('Scatter Plot: GDP vs Happiness Score (Top 25)')
plt.xlabel('Log GDP per Capita')
plt.ylabel('Happiness Score')
plt.show()
```

```python
#Heatmaps
print("Mithunjai.E\t\tDSV - 29_01_26\t\t24BAD071")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/29-01-2026.csv").head(25)

plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr = numeric_df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap (Top 25 Countries)')
plt.show()
```
